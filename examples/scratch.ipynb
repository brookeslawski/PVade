```python
#studying tau's behavior (Codina 2000, Donea & Huerta 2003)
from matplotlib.colors import LogNorm

u_mag = np.linspace(0.01,20.0,100)
h = np.linspace(0.001,1.0,100)

X, Y = np.meshgrid(u_mag,h)

alpha = 2.25E-05

Pe = np.zeros((len(u_mag),len(h)))
tau = np.zeros((len(u_mag),len(h)))

for i in range(len(u_mag)):
    for j in range(len(h)):
        Pe[i,j] = u_mag[i]*h[j]/(2*alpha)
        tau[i,j] = (h[j]/(2*u_mag[i]))*(1+(1/Pe[i,j]))**(-1)

fig,axs = plt.subplots(1, 2, figsize=[10,4], dpi=300)
c1 = axs[0].contourf(X, Y, Pe)
fig.colorbar(c1, ax=axs[0])
axs[0].set_title('Pe')
axs[0].set_xlabel('u')
axs[0].set_ylabel('h')

c2 = axs[1].contourf(X, Y, tau, levels=60, norm=LogNorm(vmin=tau.min(), vmax=tau.max()))
fig.colorbar(c2, ax=axs[1])
axs[1].set_title('tau')
axs[1].set_xlabel('u')
axs[1].set_ylabel('h')

plt.tight_layout()
```
